# Correlations

- Calls vs. Methods vs. Classes

```python
# Get columns "Calls", "DISTINCT methods" and "DISTINCT classes" from df_header and remove NaN rows
calls = df_header['Calls'].dropna().astype(int).tolist()
distinct_methods = df_header['DISTINCT methods'].dropna().astype(int).tolist()
distinct_classes = df_header['DISTINCT classes'].dropna().astype(int).tolist()

# Check if all lists have the same length
if not (len(calls) == len(distinct_methods) == len(distinct_classes)):
    raise ValueError("Columns 'Calls', 'DISTINCT methods' and 'DISTINCT classes' must have the same number of non-NaN entries.")

# Caluclate correlation matrix 
data = {
    'Calls': calls,
    'DISTINCT methods': distinct_methods,
    'DISTINCT classes': distinct_classes
}
df_corr = pd.DataFrame(data)
corr_matrix = df_corr.corr()

# Visualize correlation matrix with plotly
import plotly.express as px
fig_corr = px.imshow(
    corr_matrix,
    text_auto=".2f",
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1,
    title="Correlation Matrix"
)
fig_corr.show()
```

```python
!pip install scipy

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import gaussian_kde

if tuples_5d:
    df_plot = pd.DataFrame(
        tuples_5d,
        columns=[
            "Calls",
            "DISTINCT methods",
            "DISTINCT classes",
            "Class correct",
            "Method correct",
        ],
    )

    calls_all = df_plot["Calls"].values.astype(float)
    calls_method_false = df_plot.loc[df_plot["Method correct"] == 0, "Calls"].values.astype(float)
    calls_class_false = df_plot.loc[df_plot["Class correct"] == 0, "Calls"].values.astype(float)

    x_vals_full = np.linspace(calls_all.min(), calls_all.max(), 500)
    x_vals = x_vals_full[x_vals_full <= 5000]

    kde_all = gaussian_kde(calls_all, bw_method=0.3)
    kde_method_false = gaussian_kde(calls_method_false, bw_method=0.3)
    kde_class_false = gaussian_kde(calls_class_false, bw_method=0.3)

    all_vals_full = kde_all(x_vals_full)
    method_false_vals_full = kde_method_false(x_vals_full)
    class_false_vals_full = kde_class_false(x_vals_full)

    # Only consider x values up to 5000 for error rate estimation
    all_vals = all_vals_full[x_vals_full <= 5000]
    method_false_vals = method_false_vals_full[x_vals_full <= 5000]
    class_false_vals = class_false_vals_full[x_vals_full <= 5000]

    # Error rate estimation
    method_error_rate = np.divide(method_false_vals, all_vals, out=np.zeros_like(all_vals), where=all_vals>1e-8)
    class_error_rate = np.divide(class_false_vals, all_vals, out=np.zeros_like(all_vals), where=all_vals>1e-8)

    # Sample Size Histogram, only for calls <= 5000 to match error rate estimation range
    sample_sizes, bins = np.histogram(calls_all[calls_all <= 5000], bins=30)
    bin_centers = 0.5 * (bins[:-1] + bins[1:])

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), gridspec_kw={'height_ratios':[4,1]}, sharex=True)

    ax1.plot(x_vals, method_error_rate, color="#f97316", label="Method False Rate")
    ax1.plot(x_vals, class_error_rate, color="#0ea5e9", label="Class False Rate")
    ax1.fill_between(x_vals, 0, method_error_rate, color="#f97316", alpha=0.3)
    ax1.fill_between(x_vals, 0, class_error_rate, color="#0ea5e9", alpha=0.3)
    ax1.set_ylabel("Estimated Error Rate")
    ax1.legend()
    ax1.grid(alpha=0.3)

    ax2.bar(bin_centers, sample_sizes, width=bins[1]-bins[0], color="gray", alpha=0.5)
    ax2.set_ylabel("Sample Size")
    ax2.set_xlabel("Calls")
    ax2.grid(alpha=0.2)

    plt.tight_layout()
    plt.show()

else:
    print("No valid data found for KDE and error rate estimation.")
```

# Quality per project

```python
# Extract and plot Calls vs quality per project
import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

df_proj = df_header[['Projekt', 'Calls', 'Class correct', 'Method correct']].dropna().copy()
df_proj['class_bool'] = df_proj['Class correct'].astype(str).str.strip().str.upper().eq('T').astype(int)
df_proj['method_bool'] = df_proj['Method correct'].astype(str).str.strip().str.upper().eq('T').astype(int)

def add_series(ax, x, y, marker, color):
    jitter = rng.uniform(-0.03, 0.03, size=len(x))
    ax.scatter(x, y + jitter, marker=marker, color=color, alpha=0.65)
    if len(x) >= 2:
        coeffs = np.polyfit(x, y, 1)
        x_line = np.linspace(x.min(), x.max(), 50)
        ax.plot(x_line, coeffs[0] * x_line + coeffs[1], color=color, linestyle='--', linewidth=1.2)

if not df_proj.empty:
    projects = sorted(df_proj['Projekt'].unique())
    n = len(projects)
    cols = 3
    rows = math.ceil(n / cols)
    fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 3.6 * rows), sharey=True)
    axes = np.atleast_2d(axes)
    method_color, class_color = '#0ea5e9', '#f97316'
    rng = np.random.default_rng(seed=42)

    for idx, proj in enumerate(projects):
        r, c = divmod(idx, cols)
        g = df_proj[df_proj['Projekt'] == proj]
        x = g['Calls'].astype(int).to_numpy()
        y_cls = g['class_bool'].to_numpy()
        y_mth = g['method_bool'].to_numpy()
        add_series(axes[r, c], x, y_mth, 'o', method_color)
        add_series(axes[r, c], x, y_cls, '^', class_color)
        axes[r, c].set_title(proj)
        axes[r, c].set_xlabel('Calls')
        axes[r, c].set_ylim(-0.2, 1.2)
        axes[r, c].set_yticks([0, 1])
        axes[r, c].set_yticklabels(['F', 'T'])
        axes[r, c].grid(alpha=0.2)

    for j in range(n, rows * cols):
        r, c = divmod(j, cols)
        axes[r, c].axis('off')

    legend_handles = [
        Line2D([0], [0], marker='o', color='w', markerfacecolor=method_color, markersize=8, label='Method correct'),
        Line2D([0], [0], marker='^', color='w', markerfacecolor=class_color, markersize=8, label='Class correct'),
    ]
    fig.legend(handles=legend_handles, loc='upper center', ncol=2, bbox_to_anchor=(0.5, 1.02))
    plt.tight_layout()
    plt.show()
else:
    print('No valid data found for project-wise analysis.')
```

```python

# Visualize LLM-Patched distances and their occurrences

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

## Extract Method hops (LLM <-> patched), exclude 0 and negative values
method_hops = (
    df_header['Method hops (LLM <-> patched)']
    .dropna()
    .astype(int)
    .loc[lambda s: s > 0]
    .tolist()
)

## Extract Class hops (LLM <-> patched), exclude 0 and negative values
class_hops = (
    df_header['Class Hops (LLM <-> patched)']
    .dropna()
    .astype(int)
    .loc[lambda s: s > 0]
    .tolist()
)

method_counts = pd.Series(method_hops).value_counts().sort_index()
class_counts = pd.Series(class_hops).value_counts().sort_index()

all_indices = sorted(set(method_counts.index) | set(class_counts.index))
method_vals = [method_counts.get(i, 0) for i in all_indices]
class_vals = [class_counts.get(i, 0) for i in all_indices]

method_total = sum(method_vals)
class_total = sum(class_vals)

x = np.arange(len(all_indices))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 3.5))
bars_method = ax.bar(x - width/2, method_vals, width, color="#ff7f0e", alpha=0.85, label="Sub-component hops (Predicted ↔ Patched)")
bars_class  = ax.bar(x + width/2, class_vals,  width, color="#1f77b4", alpha=0.85, label="Super-component hops (Predicted ↔ Patched)")

# Add percentage labels on top of each bar
for bar, val in zip(bars_method, method_vals):
    if val > 0:
        pct = val / method_total * 100
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.1,
                f'{pct:.0f}%', ha='center', va='bottom', fontsize=8, color="#ff7f0e")

for bar, val in zip(bars_class, class_vals):
    if val > 0:
        pct = val / class_total * 100
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.1,
                f'{pct:.0f}%', ha='center', va='bottom', fontsize=8, color="#1f77b4")

ax.set_xlabel("Distance (Predicted ↔ Patched)")
ax.set_ylabel("Number of cases")
ax.set_title("Distribution of distances (Predicted ↔ Patched)")
ax.set_xticks(x)
ax.set_xticklabels(all_indices)
ax.legend()
ax.grid(axis='y', alpha=0.3)
ax.margins(y=0.15)
plt.tight_layout()
plt.show()

# Keep hops_counts for downstream cells
hops_counts = method_counts
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import gaussian_kde


def plot_false_rate_vs_calls(
    tuples_5d,
    *,
    max_calls: float = 5000,
    n_grid: int = 500,
    bw: float = 0.3,
    bins: int = 30,
    method_color: str = "#f97316",
    class_color: str = "#0ea5e9",
):
    """Plot KDE-based false-rate estimate vs calls + sample size histogram.

    Expects tuples_5d rows with columns:
    (Calls, DISTINCT methods, DISTINCT classes, Class correct, Method correct)
    where correctness is 1/0 (or True/False).
    """

    if not tuples_5d:
        raise ValueError("tuples_5d is empty; nothing to plot.")

    df_plot = pd.DataFrame(
        tuples_5d,
        columns=[
            "Calls",
            "DISTINCT methods",
            "DISTINCT classes",
            "Class correct",
            "Method correct",
        ],
    )

    calls_all = pd.to_numeric(df_plot["Calls"], errors="coerce").dropna().to_numpy(dtype=float)
    if calls_all.size < 2:
        raise ValueError("Not enough valid 'Calls' values for KDE.")

    calls_method_false = pd.to_numeric(
        df_plot.loc[df_plot["Method correct"].astype(int) == 0, "Calls"], errors="coerce"
    ).dropna().to_numpy(dtype=float)
    calls_class_false = pd.to_numeric(
        df_plot.loc[df_plot["Class correct"].astype(int) == 0, "Calls"], errors="coerce"
    ).dropna().to_numpy(dtype=float)

    x_vals_full = np.linspace(float(np.min(calls_all)), float(np.max(calls_all)), int(n_grid))
    x_mask = x_vals_full <= max_calls
    x_vals = x_vals_full[x_mask]

    kde_all = gaussian_kde(calls_all, bw_method=bw)
    all_vals_full = kde_all(x_vals_full)

    if calls_method_false.size >= 2:
        kde_method_false = gaussian_kde(calls_method_false, bw_method=bw)
        method_false_vals_full = kde_method_false(x_vals_full)
    else:
        method_false_vals_full = np.zeros_like(all_vals_full)

    if calls_class_false.size >= 2:
        kde_class_false = gaussian_kde(calls_class_false, bw_method=bw)
        class_false_vals_full = kde_class_false(x_vals_full)
    else:
        class_false_vals_full = np.zeros_like(all_vals_full)

    all_vals = all_vals_full[x_mask]
    method_false_vals = method_false_vals_full[x_mask]
    class_false_vals = class_false_vals_full[x_mask]

    method_error_rate = np.divide(
        method_false_vals,
        all_vals,
        out=np.zeros_like(all_vals),
        where=all_vals > 1e-8,
    )
    class_error_rate = np.divide(
        class_false_vals,
        all_vals,
        out=np.zeros_like(all_vals),
        where=all_vals > 1e-8,
    )

    # Sample Size Histogram (limit to max_calls to match the estimation range)
    sample_sizes, bin_edges = np.histogram(calls_all[calls_all <= max_calls], bins=int(bins))
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    fig, (ax1, ax2) = plt.subplots(
        2,
        1,
        figsize=(6, 5),
        gridspec_kw={"height_ratios": [4, 1]},
        sharex=True,
    )

    ax1.plot(x_vals, method_error_rate, color=method_color, label="Sub-component KDE ratio")
    ax1.plot(x_vals, class_error_rate, color=class_color, label="Super-component KDE ratio")
    ax1.fill_between(x_vals, 0, method_error_rate, color=method_color, alpha=0.3)
    ax1.fill_between(x_vals, 0, class_error_rate, color=class_color, alpha=0.3)
    ax1.set_ylabel("Relative error density (KDE ratio)")
    ax1.legend(loc="upper left")
    ax1.grid(alpha=0.3)

    ax2.bar(bin_centers, sample_sizes, width=bin_edges[1] - bin_edges[0], color="gray", alpha=0.5)
    ax2.set_ylabel("Sample Size")
    ax2.set_xlabel("Calls")
    ax2.grid(alpha=0.2)

    plt.tight_layout()
    return fig, (ax1, ax2)


# Usage: run the cells that build `tuples_5d`, then call:
#   plot_false_rate_vs_calls(tuples_5d)
#   plt.show()
if "tuples_5d" in globals():
    plot_false_rate_vs_calls(tuples_5d)
    plt.show()
else:
    print("`tuples_5d` not found. Run the cells that build it, then call plot_false_rate_vs_calls(tuples_5d).")
```

```python
import pandas as pd
import numpy as np

# Load data, skip descriptive row, remove entirely empty rows
url = "https://docs.google.com/spreadsheets/d/1WYs9NuY2V4mdaYyD1a7c9ffCQEq67qFwK_9aUvGfUnk/export?format=csv"
df_raw = pd.read_csv(url, skiprows=1).dropna(how='all')

# Base columns for primary model entries
cols_main = ["Nr", "Calls", "Class correct", "Method correct"]

def extract_model_data(df, model_search, extra_cols=None):
    """Extracts performance data for a given model, handling both primary and secondary column layouts."""
    is_main = df["Used model"].str.contains(model_search, case=False, na=False)
    df_model = df[is_main][cols_main].copy()
    
    if extra_cols:
        # Check if the model was evaluated alongside another model
        has_extra = df[extra_cols[-2:]].notna().any(axis=1) & (~is_main)
        df_extra = df.loc[has_extra, extra_cols].copy()
        df_extra.columns = cols_main
        df_model = pd.concat([df_model, df_extra], ignore_index=True)
        
    # Clean output: drop rows missing target stats and cast 'Nr'
    df_model = df_model[df_model[["Class correct", "Method correct"]].replace(r'^\s*$', np.nan, regex=True).notna().any(axis=1)]
    df_model['Nr'] = pd.to_numeric(df_model['Nr'], errors='coerce')
    
    return df_model

# 1. Extract sets
df_gpt5 = extract_model_data(df_raw, "gpt-5")
df_gemini = extract_model_data(df_raw, "Gemini", ["Nr", "Calls", "Class Correct GE", "Method Correct GE"])
df_grok = extract_model_data(df_raw, "grok", ["Nr", "Calls", "Class Correct GR", "Method Correct GR"])

# 2. Identify strictly overlapping experiments
common_nrs = set(df_gpt5['Nr'].dropna()) & set(df_gemini['Nr'].dropna()) & set(df_grok['Nr'].dropna())

# 3. Filter to overlaps
df_gpt5_overlap = df_gpt5[df_gpt5['Nr'].isin(common_nrs)].copy()
df_gemini_overlap = df_gemini[df_gemini['Nr'].isin(common_nrs)].copy()
df_grok_overlap = df_grok[df_grok['Nr'].isin(common_nrs)].copy()

print(f"GPT-5 rows: {len(df_gpt5)}, Gemini rows: {len(df_gemini)}, Grok rows: {len(df_grok)}")
print(f"Overlapping rows across all 3 models: {len(common_nrs)}")
```

```python
# Calculate the overlapping avg accuracy for the 3 models, i.e.,
# GPT-5 vs Gemini and Grok 
# Gemini vs Grok
# Get the original model experiment numbers (Nr) for each model by substracting the overlapping ones from the total sets
print("Calculating average accuracies for overlapping and non-overlapping experiments...")
nrs_gpt5 = {int(x) for x in nrs_gpt5}
nrs_gemini = {int(x) for x in nrs_gemini}
nrs_grok = {int(x) for x in nrs_grok}
nrs_gpt5_only = nrs_gpt5 - nrs_gemini
nrs_gemini_only = nrs_gemini - nrs_gpt5
nrs_grok_only = nrs_grok - nrs_gemini

# Filter the DataFrames to get the respective experiments and convert Nr to numeric for filtering
gpt5_experiments_gpt5 = df_gpt5[df_gpt5['Nr'].isin(nrs_gpt5_only)]
gpt5_experiments_gemini = df_gemini[df_gemini['Nr'].isin(nrs_gpt5)]
gpt5_experiments_grok = df_grok[df_grok['Nr'].isin(nrs_gpt5)]

gemini_experiments_gemini = df_gemini[df_gemini['Nr'].isin(nrs_gemini_only)]
gemini_experiments_grok = df_grok[df_grok['Nr'].isin(nrs_gemini)]

grok_experiments_grok = df_grok[df_grok['Nr'].isin(nrs_grok_only)]

# Average all experiments
def calculate_avg_accuracy(df, model_name):
    class_acc = df['Class correct'].astype(str).str.strip().str.upper().eq('T').mean()
    method_acc = df['Method correct'].astype(str).str.strip().str.upper().eq('T').mean()
    print(f"{model_name} - Class Accuracy: {class_acc:.2%}, Method Accuracy: {method_acc:.2%}, Sample Size: {len(df)}")

calculate_avg_accuracy(gpt5_experiments_gpt5, "GPT-5 (original)")
calculate_avg_accuracy(gpt5_experiments_gemini, "GPT-5 (Gemini overlap)")
calculate_avg_accuracy(gpt5_experiments_grok, "GPT-5 (Grok overlap)")
calculate_avg_accuracy(gemini_experiments_gemini, "Gemini (original)")
calculate_avg_accuracy(gemini_experiments_grok, "Gemini (Grok overlap)")
calculate_avg_accuracy(grok_experiments_grok, "Grok (original)")
```

```python
# Get the overall average accuracy across all experiments for each model and the precise number of correct classes/methods and total samples
def calculate_overall_accuracy(df, model_name):
    class_correct = df['Class correct'].astype(str).str.strip().str.upper().eq('T').sum()
    method_correct = df['Method correct'].astype(str).str.strip().str.upper().eq('T').sum()
    total_samples = len(df)
    class_acc = class_correct / total_samples if total_samples > 0 else 0
    method_acc = method_correct / total_samples if total_samples > 0 else 0
    print(f"{model_name} - Overall Class Accuracy: {class_acc:.2%} ({class_correct}/{total_samples}), Overall Method Accuracy: {method_acc:.2%} ({method_correct}/{total_samples})")  
calculate_overall_accuracy(df_gpt5, "GPT-5 (overall)")
calculate_overall_accuracy(df_gemini, "Gemini (overall)")
calculate_overall_accuracy(df_grok, "Grok (overall)")
```

```python
from SPARQLWrapper import SPARQLWrapper, JSON, TURTLE
import xml.etree.ElementTree as ET

SPARQL_ENDPOINT = "http://localhost:8890/sparql"  # Beispiel-Endpoint, anpassen je nach Setup
sparql = SPARQLWrapper(SPARQL_ENDPOINT)
sparql.setReturnFormat(JSON)
# Function that returns the XML content instead of writing to file (optional), s.t. I can call the method from elsewhere with the graph and get the XML string
def get_hierarchy_xml_string(graph_uri: str) -> str:
    sparql.setQuery(f"""
    PREFIX ex: <http://example.org/>
    PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

    SELECT ?method ?name ?callee ?argType ?argValue ?resType ?resValue
    FROM <{graph_uri}>
    WHERE {{
        ?method ex:method ?name ;
                ex:callee ?callee .

        OPTIONAL {{
            ?method ex:args ?argList .
            ?argList rdf:rest*/rdf:first ?argNode .
            ?argNode rdf:type ?argType ;
                     rdf:value ?argValue .
        }}

        OPTIONAL {{
            ?method ex:result ?resNode .
            ?resNode rdf:type ?resType ;
                     rdf:value ?resValue .
        }}
    }}
    """)

    results = sparql.query().convert()

    # --- Methoden-Dict aufbauen (aggregiert args und single result) ---
    methods = {}
    for row in results["results"]["bindings"]:
        method_id = row["method"]["value"]
        # Name und callee könnten in manchen DBs fehlen – defensiv behandeln
        name = row["name"]["value"] if "name" in row else ""
        callee = row["callee"]["value"] if "callee" in row else method_id

        if method_id not in methods:
            methods[method_id] = {
                "id": method_id,
                "name": name,
                "callee": callee,
                "args": [],        # list of {"type":..., "value":...}
                "result": None,    # single {"type":..., "value":...} or None
                "children": []     # list of method ids called by this method
            }
        else:
            # Falls Name/Callee in späteren Zeilen nicht gesetzt waren, aktualisieren
            if not methods[method_id]["name"] and name:
                methods[method_id]["name"] = name
            if not methods[method_id]["callee"] and callee:
                methods[method_id]["callee"] = callee

        # args (können mehrere Zeilen erzeugen)
        if "argType" in row and "argValue" in row:
            methods[method_id]["args"].append({
                "type": row["argType"]["value"],
                "value": row["argValue"]["value"]
            })
        # result (falls mehrfach wegen SPARQL-Duplikaten auftaucht, überschreiben wir nicht - behalten erstes)
        if methods[method_id]["result"] is None and "resType" in row and "resValue" in row:
            methods[method_id]["result"] = {
                "type": row["resType"]["value"],
                "value": row["resValue"]["value"]
            }
    # --- Kinder (aufgerufene Methoden) zuweisen ---
    for m in methods.values():
        if m["id"] != m["callee"]:  # selbstaufruf ignorieren bei Zuordnung
            if m["callee"] in methods:
                methods[m["callee"]]["children"].append(m["id"])
    # --- Root finden (callee == id) ---
    root = next((m for m in methods.values() if m["id"] == m["callee"]), None)
    if root is None:
        raise ValueError("Kein Root gefunden (keine Methode mit ex:callee == id)")
    # --- XML Builder mit Zyklus-Schutz ---
    def build_xml_node(method_id, visited=None):
        if visited is None:
            visited = set()
        if method_id in visited:
            # Zyklus erkannt: um Endlosschleife zu vermeiden, referenzieren wir kurz und beenden
            ref = ET.Element("methodRef", attrib={"id": method_id})
            return ref

        visited.add(method_id)
        m = methods[method_id]
        elem = ET.Element("method", attrib={"id": m["id"]})
        if m["name"]:
            elem.set("name", m["name"])

        # args einbetten
        if m["args"]:
            args_elem = ET.SubElement(elem, "args")
            for arg in m["args"]:
                arg_elem = ET.SubElement(args_elem, "arg", attrib={"type": arg["type"]})
                arg_elem.text = arg["value"]

        # result (einzelnes Element, falls vorhanden)
        if m["result"]:
            res = m["result"]
            res_elem = ET.SubElement(elem, "result", attrib={"type": res["type"]})
            res_elem.text = res["value"]

        # methods (aufgerufene Methoden)
        methods_elem = ET.SubElement(elem, "methods")
        for child_id in m["children"]:
            child_node = build_xml_node(child_id, visited.copy())  # copy visited für getrennte Pfade
            methods_elem.append(child_node)

        return elem
    # --- XML-Dokument erstellen und als String zurückgeben ---
    root_elem = build_xml_node(root["id"])
    indent(root_elem)  # optional: schön formatieren
    xml_string = ET.tostring(root_elem, encoding="utf-8", method="xml").decode("utf-8")
    return xml_string

def indent(elem, level=0):
    i = "\n" + level * "  "
    if len(elem):
        if not elem.text or not elem.text.strip():
            elem.text = i + "  "
        for child in elem:
            indent(child, level + 1)
        if not child.tail or not child.tail.strip():
            child.tail = i
    if level and (not elem.tail or not elem.tail.strip()):
        elem.tail = i
    return elem
```

```python
# Plot Calls vs. Tokens
# For that, we need to create the prompts for each graph and calculate the tokens for each prompt. We can use tiktoken for that.
import pandas as pd
import tiktoken

# Load data
data_path = "../data/sheet_smart_with_failure_with_exception_chain_with_baseline_patch_check.json" 
data = pd.read_json(data_path)
prompt_template = open("../prompts/evaluation_prompt.txt", "r", encoding="utf-8").read()

calls_and_tokens = []

# For each object in the data, get the graph URI, create the prompt and calculate the tokens
for item in data.itertuples():
    graph_uri = item.Graph
    call_tree_xml = get_hierarchy_xml_string(graph_uri=graph_uri)
    prompt = prompt_template.replace("{calltree_xml}", call_tree_xml)
    encoding = tiktoken.get_encoding("cl100k_base")
    tokens = encoding.encode(prompt)
    calls = item.Metadata["Calls"]
    calls_and_tokens.append((len(tokens), calls))
    print(f"Graph URI: {graph_uri}, Tokens: {len(tokens)}, Calls: {calls}")

# Now we can plot the calls vs. tokens using matplotlib
import matplotlib.pyplot as plt
tokens, calls = zip(*calls_and_tokens)
plt.figure(figsize=(8, 5))
plt.scatter(tokens, calls, color='#0ea5e9', alpha=0.7)
plt.xlabel("Number of Tokens in Prompt")
plt.ylabel("Number of Calls in Call Tree")
plt.title("Calls vs. Tokens for Call Tree Prompts")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

```python
# Plot again with a regression line
import matplotlib.pyplot as plt
import numpy as np

tokens, calls = zip(*calls_and_tokens)
plt.figure(figsize=(8, 5))
plt.scatter(tokens, calls, color='#0ea5e9', alpha=0.7)
plt.xlabel("Number of Tokens in Prompt")
plt.ylabel("Number of Calls in Call Tree")
plt.title("Calls vs. Tokens for Call Tree Prompts")
plt.grid(alpha=0.3)
# Add regression line
z = np.polyfit(tokens, calls, 1)
p = np.poly1d(z)
plt.plot(tokens, p(tokens), color='#f97316', linewidth=2, label='Regression Line')
plt.legend()
plt.tight_layout()
plt.show()
```
